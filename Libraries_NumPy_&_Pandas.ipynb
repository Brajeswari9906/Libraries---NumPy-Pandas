```python
import pandas as pd

df = pd.read_csv("retail_dataset.csv")

print(df.head())

print(df.info())
```

Q2. Inspect the dataset and identify missing values in each column. Display the total number of missing values present in the dataset.

```python
print(df.isnull().sum())

print(df.isnull().sum().sum())
```

Q3. Replace missing values in the Quantity and Price columns with their respective mean values using Pandas.

```python
df['Quantity'] = df['Quantity'].fillna(df['Quantity'].mean())

df['Price'] = df['Price'].fillna(df['Price'].mean())

print(df[['Quantity', 'Price']].head())
```

Q4. Remove all rows where the Product Category or Region column contains missing values.

```python
df = df.dropna(subset=['Product Category', 'Region'])
```

Rows with missing values in Product Category and Region were removed.

Q5. Create a new column called Revenue using NumPy.

```python
import numpy as np

df['Revenue'] = np.array(df['Quantity'] * df['Price'])

print(df.head())
```

Q6. Calculate the total revenue generated from all transactions.

```python
total_revenue = np.sum(df['Revenue'])

print(total_revenue)
```

Q7. Group the dataset by Product Category and calculate total revenue.

```python
category_revenue = df.groupby('Product Category')['Revenue'].sum()

print(category_revenue)
```

Q8. Identify top 3 and bottom 3 categories based on revenue.

```python
top_3 = category_revenue.sort_values(ascending=False).head(3)

bottom_3 = category_revenue.sort_values().head(3)

print(top_3)
print(bottom_3)
```

Q9. Group dataset by Region and calculate total revenue.

```python
region_revenue = df.groupby('Region')['Revenue'].sum()

print(region_revenue)
```

Q10. Extract month from Date column and calculate monthly revenue. Also calculate mean, median, and standard deviation of Revenue.

```python
df['Date'] = pd.to_datetime(df['Date'], dayfirst=True)

df['Month'] = df['Date'].dt.month_name()

monthly_revenue = df.groupby('Month')['Revenue'].sum()

print(monthly_revenue)

mean_revenue = np.mean(df['Revenue'])
median_revenue = np.median(df['Revenue'])
std_revenue = np.std(df['Revenue'])

print(mean_revenue)
print(median_revenue)
print(std_revenue)
```
